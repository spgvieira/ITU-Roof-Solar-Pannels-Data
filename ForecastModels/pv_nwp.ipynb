# Data

```python
from pv_nwp import *

CHANNELS = [
    "temperature_2m",
    "shortwave_radiation",
    "direct_radiation",
    "cloud_cover_low",
    "cloud_cover_mid",
    "cloud_cover_high",
    "cloud_cover",
]

START_DATE = dt.datetime(2025, 10, 15, 0, 0)
END_DATE = dt.datetime(2026, 9, 1, 0, 0)

PV_CSV = '../Datasets/PV.csv'
NWP_CSV = '../Datasets/ECMWF_forecast.csv'

dataset = load_dataset(NWP_CSV, PV_CSV, START_DATE, END_DATE)
```

Correlation between the different input variables

```python
correlation_heatmap(dataset)
```

Plot metrics against solar power output

```python
for c in CHANNELS:
    plot_nwp_vs_pv_data(dataset, c)
```

```python
for c in CHANNELS:
    box_plot(dataset, c)

box_plot(dataset, 'solar_power')
```

```python
for c in CHANNELS:
    scatter_plot(dataset, c)
```

```python
for c in CHANNELS:
    z_score(dataset, c)

z_score(dataset, 'solar_power')
```
